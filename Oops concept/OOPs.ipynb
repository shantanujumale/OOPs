What is OOP?
OOP stands for Object-Oriented Programming.

It is a programming approach where we organize our code using:
- Classes
- Objects
- Attributes
- Methods

Instead of writing all the code in separate functions, we create objects that contain both data and functionality.

```python
class car:

  def __init__(self,brand,color):

    self.brand = brand
    self.color = color

  def start(self):
    print("car is starting")

  def stop(self):
    print("car is stopped0")
```

```python
car1 = car("bmw",'blue')
```

```python
print(car1.brand)
print(car1.color)
```

```python
print(car1.start())
```
